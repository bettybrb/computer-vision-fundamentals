# Convolution and Image Filtering

This notebook explores spatial image filtering using two-dimensional convolution.

Experiments investigate smoothing, averaging, Gaussian-like filtering and Laplacian edge detection using different kernel configurations.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Create output folder (flat, simple)
os.makedirs("outputs", exist_ok=True)

# Load ONE image from DatasetA (you can change filename if you want)
img = cv2.imread("DatasetA/car-1.jpg")   # coursework says use Dataset A
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("Task 2 – Original Image")
plt.show()
```

```python
def conv2D(I, K):
    kh, kw = K.shape
    pad_h, pad_w = kh // 2, kw // 2

    # Reflect padding (minimally required)
    Ipad = cv2.copyMakeBorder(I, pad_h, pad_h, pad_w, pad_w, cv2.BORDER_REFLECT)

    out = np.zeros_like(I, dtype=np.float32)

    # Apply convolution channel-by-channel
    for y in range(I.shape[0]):
        for x in range(I.shape[1]):
            region = Ipad[y:y+kh, x:x+kw]    # (kh,kw,3)
            out[y, x] = np.sum(region * K[:, :, None], axis=(0,1))

    return np.clip(out, 0, 255).astype(np.uint8)
```

```python
identity = np.array([[0,0,0],
                     [0,1,0],
                     [0,0,0]], np.float32)

out_2a = conv2D(img, identity)
```

```python
fig, ax = plt.subplots(1,2, figsize=(8,4))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); ax[0].axis("off"); ax[0].set_title("Original")
ax[1].imshow(cv2.cvtColor(out_2a, cv2.COLOR_BGR2RGB)); ax[1].axis("off"); ax[1].set_title("2(a) Output")
plt.show()

cv2.imwrite("outputs/2_original_img.png", out_2a)
```

```python
avg3  = np.ones((3,3), np.float32) / 9
avg5  = np.ones((5,5), np.float32) / 25
avg35 = np.ones((3,5), np.float32) / 15
```

```python
out_avg3  = conv2D(img, avg3)
out_avg5  = conv2D(img, avg5)
out_avg35 = conv2D(img, avg35)

fig, ax = plt.subplots(2,2, figsize=(10,10))
ax[0,0].imshow(cv2.cvtColor(img,       cv2.COLOR_BGR2RGB)); ax[0,0].set_title("Original"); ax[0,0].axis("off")
ax[0,1].imshow(cv2.cvtColor(out_avg3,  cv2.COLOR_BGR2RGB)); ax[0,1].set_title("3×3 Avg"); ax[0,1].axis("off")
ax[1,0].imshow(cv2.cvtColor(out_avg5,  cv2.COLOR_BGR2RGB)); ax[1,0].set_title("5×5 Avg"); ax[1,0].axis("off")
ax[1,1].imshow(cv2.cvtColor(out_avg35, cv2.COLOR_BGR2RGB)); ax[1,1].set_title("3×5 Avg"); ax[1,1].axis("off")
plt.show()

cv2.imwrite("outputs/2b_avg3.png", out_avg3)
cv2.imwrite("outputs/2b_avg5.png", out_avg5)
cv2.imwrite("outputs/2b_avg35.png", out_avg35)
```

```python
A = np.array([[1,2,1],
              [2,4,2],
              [1,2,1]], np.float32) / 16   # Gaussian-like

B = np.array([[0, 1, 0],
              [1,-4, 1],
              [0, 1, 0]], np.float32)     # Laplacian edge
```

```python
out_A = conv2D(img, A)
out_B = conv2D(img, B)
```

```python
fig, ax = plt.subplots(1,3, figsize=(12,4))
ax[0].imshow(cv2.cvtColor(img,   cv2.COLOR_BGR2RGB)); ax[0].set_title("Original"); ax[0].axis("off")
ax[1].imshow(cv2.cvtColor(out_A, cv2.COLOR_BGR2RGB)); ax[1].set_title("Kernel A"); ax[1].axis("off")
ax[2].imshow(cv2.cvtColor(out_B, cv2.COLOR_BGR2RGB)); ax[2].set_title("Kernel B"); ax[2].axis("off")
plt.show()
```

```python
cv2.imwrite("outputs/2c_A.png", out_A)
cv2.imwrite("outputs/2c_B.png", out_B)
```

```python
out_AA = conv2D(conv2D(img, A), A)
out_AB = conv2D(conv2D(img, A), B)
out_BA = conv2D(conv2D(img, B), A)
```

```python
fig, ax = plt.subplots(1,3, figsize=(12,4))
ax[0].imshow(cv2.cvtColor(out_AA, cv2.COLOR_BGR2RGB)); ax[0].set_title("A → A"); ax[0].axis("off")
ax[1].imshow(cv2.cvtColor(out_AB, cv2.COLOR_BGR2RGB)); ax[1].set_title("A → B"); ax[1].axis("off")
ax[2].imshow(cv2.cvtColor(out_BA, cv2.COLOR_BGR2RGB)); ax[2].set_title("B → A"); ax[2].axis("off")
plt.show()
```

```python
cv2.imwrite("outputs/2d_AA.png", out_AA)
cv2.imwrite("outputs/2d_AB.png", out_AB)
cv2.imwrite("outputs/2d_BA.png", out_BA)
```
